### Import the required libraries

```python
# Import libraries for data loading, plotting, and path handling.
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import warnings

# Suppress warnings
warnings.filterwarnings("ignore")

# Display settings
pd.set_option('display.max_columns', None)
sns.set_style("whitegrid")
```

### Load the data

```python
# Point to the dataset location in the project folder.
file_path = Path('DataSets_Project/nationaldatabaseofchildcareprices.xlsx')

# Read the dataset into a DataFrame.
df = pd.read_excel(file_path)

# Display basic shape and key time/geography coverage.
print('Rows:', len(df))
print('Columns:', df.shape[1])
print('Year range:', int(df['StudyYear'].min()), 'to', int(df['StudyYear'].max()))
print('Distinct states:', df['State_Name'].nunique())
print('Distinct counties:', df['County_FIPS_Code'].nunique())
```

### Field Interpretation

Based on field names and data structure in the workbook:
- Geographic keys: `State_Name`, `State_Abbreviation`, `County_Name`, `County_FIPS_Code`.
- Time key: `StudyYear` (2008-2018 in this file).
- Economic context: `MHI`, `MHI_2018`, `ME`, `FME`, `MME` (household and earnings variables).
- Labor context: `UNR_*`, `FUNR_*`, `MUNR_*`, `FLFPR_*`, `MLFPR_*` (unemployment and labor-force participation).
- Childcare prices : `MCInfant`, `MCToddler`, `MCPreschool`.
- Childcare prices (family child care homes): `MFCCInfant`, `MFCCToddler`, `MFCCPreschool`.
- Affordability proxies: `_75C*` and `_75FCC*` fields likely represent prices at the 75th percentile threshold.
- Data quality flags: columns ending in `_flag` and prefixed with `i` indicate imputation or quality markers.

```python
# Check missingness and medians for core childcare and income fields.
key_cols = ['MCInfant', 'MCToddler', 'MCPreschool', \
            'MFCCInfant', 'MHI', 'MHI_2018']

summary = pd.DataFrame({
    'missing_pct': df[key_cols].isna().mean() * 100,
    'median_value': df[key_cols].median(numeric_only=True)
}).round(2)

summary
```

```python
# Calculate price trends by year for selected childcare price metrics.
trend_cols = ['MCInfant', 'MCToddler', 'MCPreschool', 'MFCCInfant']
yearly_medians = df.groupby('StudyYear')\
    [trend_cols].median(numeric_only=True)

# Show start/end values and percent change across the time period.
changes = []
for col in trend_cols:
    start = yearly_medians[col].iloc[0]
    end = yearly_medians[col].iloc[-1]
    pct_change = (end / start - 1) * 100
    changes.append({
        'metric': col,
        'start_value': round(start, 2),
        'end_value': round(end, 2),
        'pct_change': round(pct_change, 2)
    })

pd.DataFrame(changes)
```

```python
# Find states with highest and lowest county-level median infant center prices in the latest year.
latest_year = int(df['StudyYear'].max())
latest = df[df['StudyYear'] == latest_year]
state_price = latest.groupby('State_Name')['MCInfant']\
    .median().dropna().sort_values(ascending=False)

top_5 = state_price.head(5).round(2)
bottom_5 = state_price.tail(5).round(2)

print('Latest year:', latest_year)
print('Top 5 highest states:', top_5)
print('Bottom 5 lowest states:', bottom_5)
```

### What Is Most Interesting About the Data and What I Want to Present and Draw Attention

What I find most interesting is the combined pattern of rising childcare prices and strong geographic inequality. The data shows that costs increase over time, but the burden is not shared evenly across states and counties.

The story I want to draw attention to is a three-part message:
- **Trend pressure:** Childcare prices increased notably between 2008 and 2018.
- **Affordability pressure:** Childcare cost burden rises when compared with household income.
- **Place-based inequality:** Some states and counties are consistently much more expensive than others.

This combination is important because families experience childcare access very differently based on where they live, not just based on national averages.

### Target Audience

My target audience is **state policymakers and county-level administrators** (human services, early childhood offices, and budget teams).

Why this audience:
- They are familiar with public-policy metrics but may not be deeply technical.
- They can act on findings through subsidy rules, reimbursement rates, and program prioritization.
- They need concise evidence showing where affordability pressure is highest and how trends are changing.

I want audience to understand childcare affordability should be treated as a **location-specific policy problem**, not only a national trend story.

### Trend Images

The following cells add a few visual images that highlight key data trends: childcare prices over time, affordability burden over time, and interstate variation in the latest year.

### 1: Childcare Prices Over Time

```python
trend_plot = df.groupby('StudyYear')[['MCInfant', \
    'MCToddler', 'MCPreschool']].median().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=trend_plot, x='StudyYear', \
             y='MCInfant', marker='o', label='Infant (Center)')
sns.lineplot(data=trend_plot, x='StudyYear', \
             y='MCToddler', marker='o', label='Toddler (Center)')
sns.lineplot(data=trend_plot, x='StudyYear', \
             y='MCPreschool', marker='o', label='Preschool (Center)')

plt.title('Trend 1: Median Center-Based Childcare Prices (2008-2018)')
plt.xlabel('Study Year')
plt.ylabel('Median Monthly Price (USD)')
plt.legend()
plt.tight_layout()
plt.show()
```

### 2: Affordability Burden Over Time

```python
burden_plot = df[['StudyYear', 'MCInfant', \
                  'MHI_2018']].dropna().copy()
burden_plot['burden_pct'] = burden_plot['MCInfant']\
      * 12 / burden_plot['MHI_2018'] * 100
burden_year = burden_plot.groupby('StudyYear', \
        as_index=False)['burden_pct'].median()

plt.figure(figsize=(12, 6))
sns.lineplot(data=burden_year, x='StudyYear', \
        y='burden_pct', marker='o', color='#d95f02')

plt.title('Trend 2: Median Childcare Affordability \
          Burden (Infant Center Care)')
plt.xlabel('Study Year')
plt.ylabel('Annual Cost as % of Median Household Income')
plt.tight_layout()
plt.show()
```

### 3: State-Level Cost Disparity

```python
latest_year_plot = int(df['StudyYear'].max())
state_latest = (
    df[df['StudyYear'] == latest_year_plot]
    .groupby('State_Name', as_index=False)['MCInfant']
    .median()
    .dropna()
    .sort_values('MCInfant', ascending=False)
)

plt.figure(figsize=(12, 8))
sns.barplot(data=state_latest.head(12), \
            x='MCInfant', y='State_Name', palette='viridis')

plt.title(f'Trend 3: Top 12 Highest-Cost States \
          for Infant Care ({latest_year_plot})')
plt.xlabel('Median Monthly Infant Care Price (USD)')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```
