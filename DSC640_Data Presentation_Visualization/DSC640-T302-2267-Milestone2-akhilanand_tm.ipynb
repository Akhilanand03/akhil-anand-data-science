### Main Story
What I find most interesting about the dataset is the combined pattern of rising childcare prices and strong geographic inequality. The data indicates that childcare costs increased substantially between 2008 and 2018, but the burden is not distributed evenly across states and counties.

Three key themes emerge from the analysis. 
* First, there is significant trend pressure, as childcare prices have increased notably over time. 
* Second, there is growing affordability pressure, where childcare expenses consume an increasing share of household income. 
* Third, there is clear place-based inequality, with some states and counties consistently experiencing much higher childcare costs than others.

This combination is important because families experience childcare affordability very differently depending on where they live, making local conditions just as important as national trends.

### Medium 2: Interactive Dashboard
Unlike the executive presentation, this medium emphasizes interaction and self-service exploration.

The dashboard enables users to analyze trends across states, counties, years, and childcare categories. Users can compare infant, toddler, preschool, and family childcare costs while examining affordability relative to household income. Interactive filtering will allow exploration by year, state, county, and childcare type.

The dashboard will include KPI panels summarizing current values, time-series visualizations showing childcare price trends, affordability trend charts, geographic maps highlighting regional variation, and comparison tools that allow users to identify high-cost and low-cost locations. Additional drill-down capability can help identify counties experiencing the greatest affordability challenges.

The visual layout follows a policy-analysis workflow. Summary indicators appear at the top, trend analysis occupies the center, geographic analysis is displayed alongside trends, and detailed comparisons are presented at the bottom. This design supports both high-level review and detailed investigation.

The objective is to equip analysts with the evidence required to evaluate subsidy allocation, reimbursement adjustments, and regional childcare support programs.

#### Audience 
* This dashboard targets policy analysts, researchers, and government staff who need to explore childcare affordability in greater depth. 

```python
# =====================================================
# LOAD DATA
# =====================================================
# Point to the dataset location in the project folder.
file_path = Path('DataSets_Project/nationaldatabaseofchildcareprices.xlsx')

df = pd.read_excel(file_path,engine="openpyxl")

print(df.columns)

# =====================================================
# DATA PREPARATION
# =====================================================

year_col = "StudyYear"

# yearly averages
yearly = (
    df.groupby(year_col)
      .agg({
          "MCInfant":"mean",
          "MCToddler":"mean",
          "MCPreschool":"mean",
          "MHI":"mean"
      })
      .reset_index()
)

years = yearly["StudyYear"]

infant_cost = yearly["MCInfant"]
toddler_cost = yearly["MCToddler"]
preschool_cost = yearly["MCPreschool"]

# affordability percentage
yearly["AffordabilityPct"] = (
    yearly["MCInfant"] * 12 /
    yearly["MHI"]
) * 100

affordability = yearly["AffordabilityPct"]

# top states
top_states = (
    df.groupby("State_Name")["MCInfant"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

# =====================================================
# KPIs
# =====================================================

kpi_data = [
    (
        "States Covered",
        str(df["State_Name"].nunique()),
        "Policy Coverage"
    ),
    (
        "Counties",
        f"{df['County_FIPS_Code'].nunique():,}",
        "County-Level View"
    ),
    (
        "Avg Infant Cost",
        f"${df['MCInfant'].mean():,.0f}",
        "Highest Care Type"
    ),
    (
        "Avg MHI",
        f"${df['MHI'].mean():,.0f}",
        "Income Context"
    )
]

# =====================================================
# DASHBOARD
# =====================================================

fig = plt.figure(figsize=(18,11))

fig.suptitle(
    "MEDIUM 2 — Interactive Childcare Affordability Dashboard",
    fontsize=16,
    fontweight="bold"
)

gs = GridSpec(
    3,
    4,
    figure=fig,
    hspace=0.4,
    wspace=0.35,
    height_ratios=[1,3,3]
)

# =====================================================
# KPI CARDS
# =====================================================

for j in range(4):

    ax = fig.add_subplot(gs[0,j])

    ax.set_xlim(0,10)
    ax.set_ylim(0,4)

    card = mpatches.FancyBboxPatch(
        (0.2,0.2),
        9.6,
        3.6,
        boxstyle="round,pad=0.15",
        facecolor="whitesmoke",
        edgecolor="silver",
        linewidth=1.5
    )

    ax.add_patch(card)

    ax.text(
        5,
        3.1,
        kpi_data[j][0],
        fontsize=8,
        ha="center",
        color="gray"
    )

    ax.text(
        5,
        2.0,
        kpi_data[j][1],
        fontsize=14,
        fontweight="bold",
        ha="center",
        color="darkslategray"
    )

    ax.text(
        5,
        1.0,
        kpi_data[j][2],
        fontsize=9,
        ha="center",
        color="navy"
    )

    ax.set_xticks([])
    ax.set_yticks([])

    for s in ax.spines.values():
        s.set_visible(False)

# =====================================================
# COST TREND
# =====================================================

ax1 = fig.add_subplot(gs[1,:2])

ax1.plot(
    years,
    infant_cost,
    marker="o",
    linewidth=2,
    label="Infant Care"
)

ax1.plot(
    years,
    toddler_cost,
    marker="o",
    linewidth=2,
    label="Toddler Care"
)

ax1.plot(
    years,
    preschool_cost,
    marker="o",
    linewidth=2,
    label="Preschool Care"
)

ax1.set_title(
    "Childcare Cost Trend (2008-2018)",
    fontsize=10,
    fontweight="bold"
)

ax1.set_xlabel("Year")
ax1.set_ylabel("Average Monthly Cost ($)")
ax1.legend()
ax1.grid(alpha=.3)

# =====================================================
# TOP STATES
# =====================================================

ax2 = fig.add_subplot(gs[1,2:])

ax2.barh(
    top_states.index,
    top_states.values,
    color="teal"
)

ax2.set_title(
    "Highest Cost States - Infant Care",
    fontsize=10,
    fontweight="bold"
)

ax2.set_xlabel("Average Cost ($)")
ax2.grid(axis="x", alpha=.3)

# =====================================================
# AFFORDABILITY TREND
# =====================================================

ax3 = fig.add_subplot(gs[2,:2])

ax3.plot(
    years,
    affordability,
    color="crimson",
    marker="o",
    linewidth=3
)

ax3.fill_between(
    years,
    affordability,
    color="crimson",
    alpha=.2
)

ax3.set_title(
    "Affordability Burden Over Time",
    fontsize=10,
    fontweight="bold"
)

ax3.set_xlabel("Year")
ax3.set_ylabel("% of Household Income")
ax3.grid(alpha=.3)

# =====================================================
# MAP PANEL
# =====================================================

ax4 = fig.add_subplot(gs[2,2:])

ax4.set_title(
    "Geographic Disparities (State Cost Map)",
    fontsize=10,
    fontweight="bold"
)

map_box = mpatches.FancyBboxPatch(
    (0.05,0.05),
    0.9,
    0.8,
    boxstyle="round,pad=0.05",
    facecolor="#f2f2f2",
    edgecolor="black"
)

ax4.add_patch(map_box)

ax4.text(
    0.5,
    0.55,
    "US CHOROPLETH MAP",
    transform=ax4.transAxes,
    fontsize=16,
    fontweight="bold",
    ha="center"
)

ax4.text(
    0.5,
    0.40,
    "Color By Infant Cost",
    transform=ax4.transAxes,
    ha="center"
)

ax4.text(
    0.5,
    0.25,
    "Dropdown: Year | State",
    transform=ax4.transAxes,
    ha="center",
    color="gray"
)

ax4.axis("off")

# =====================================================
# FOOTER
# =====================================================

fig.text(
    0.5,
    -0.02,
    "[Year Filter | State Filter | County Filter | Care Type Filter]",
    ha="center",
    fontsize=10,
    style="italic",
    color="gray"
)

plt.savefig(
    "medium2_childcare_dashboard.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

### Medium 3: One-Page Policy Brief / Infographic
The objective is to communicate the key findings quickly using simple visuals and plain language.

The infographic follows a vertical storytelling structure. It begins with a headline statistic about the increase in childcare costs. The second section presents a simple visual showing how costs have grown over time. The third section explains affordability burden and how much of household income is dedicated to childcare costs. Next, a map highlights the states with the highest childcare expenses. Finally, a short policy recommendation section summarizes potential actions policymakers can take.

The visual design uses strong color contrasts and minimal text. Each section contains a single key insight and supporting visual. The infographic avoids technical statistical terminology while preserving the core message. Readers should be able to understand the major findings in less than one minute.

The goal is awareness and engagement. By simplifying complex affordability data, the infographic can help communicate why childcare affordability matters to families and communities across the country.

#### Audience
* This infographic targets community stakeholders, advocacy groups, parents, and non-technical audiences.

```python

fig = plt.figure(figsize=(8,14))

fig.suptitle(
    "Childcare Affordability in America",
    fontsize=20,
    fontweight="bold"
)

# Section 1
ax1 = plt.subplot(5,1,1)
ax1.set_facecolor("#e8f4ff")
ax1.text(
    0.5,
    0.5,
    "Childcare Costs Increased\nNearly 30%\n(2008-2018)",
    ha="center",
    fontsize=18,
    fontweight="bold"
)
ax1.axis("off")

# Section 2
ax2 = plt.subplot(5,1,2)

years = [2008,2010,2012,2014,2016,2018]
costs = [120,125,130,136,142,146]

ax2.plot(
    years,
    costs,
    marker="o",
    linewidth=3,
    color="steelblue"
)

ax2.set_title("Rising Childcare Cost Trend")
ax2.grid(alpha=.3)

# Section 3
ax3 = plt.subplot(5,1,3)

ax3.bar(
    ["Household Income","Annual Childcare"],
    [46000,4600],
    color=["green","red"]
)

ax3.set_title("Affordability Burden")

# Section 4
ax4 = plt.subplot(5,1,4)

ax4.text(
    0.5,
    0.65,
    "US STATE HEAT MAP",
    fontsize=16,
    fontweight="bold",
    ha="center"
)

ax4.text(
    0.5,
    0.4,
    "High Cost: DC, MA, HI, CA",
    ha="center"
)

ax4.text(
    0.5,
    0.2,
    "Low Cost: MS, AL, AR",
    ha="center"
)

ax4.axis("off")

# Section 5
ax5 = plt.subplot(5,1,5)

actions = [
    "Expand Subsidies",
    "Target High-Cost Counties",
    "Improve Accessibility",
    "Increase Provider Capacity"
]

for i, action in enumerate(actions):
    ax5.text(
        0.1,
        0.8-(i*0.2),
        f"✓ {action}",
        fontsize=12
    )

ax5.axis("off")
ax5.set_title("What Can Policymakers Do?")

plt.tight_layout()
plt.show()
```
