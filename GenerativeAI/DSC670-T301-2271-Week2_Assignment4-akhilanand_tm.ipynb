### Step1 : Extract Text from PDF
* The PyPDF2 library was used to read the invoice document and extract the text from each page. The extracted text was stored in a variable named invoice_text.
* This method was selected because the invoice contained machine-readable text rather than scanned images,

```python
from PyPDF2 import PdfReader

reader = PdfReader("dsc-670-exercise-invoice.pdf")

invoice_text = ""

for page in reader.pages:
    invoice_text += page.extract_text() + "\n"

print(invoice_text)
```

### Step2: Convert Text to JSON Using GPT-4.1
* The OpenAI API was used to convert the unstructured content into structured JSON.
* The prompt explicitly instructed the model to return only valid JSON. This prompt engineering technique reduced ambiguity and improved the consistency of the generated output.

```python
from openai import OpenAI

client = OpenAI(api_key="YOUR_API_KEY")  

response = client.responses.create(
    model="gpt-4.1",
    input=f"""
Convert the following invoice to structured JSON.

Invoice:

{invoice_text}

Return only valid JSON.
"""
)

print(response.output_text)
```

### Results and Evaluation

The extraction process successfully captured the invoice information, including the invoice number, purchase order number, vendor information, billing information, line items, credit amount, tax percentage, discounts, and balance due. The LLM then interpreted this information and organized it into a hierarchical JSON structure.

One advantage of this approach is that GPT-4.1 was able to recognize relationships between invoice fields without requiring complex parsing logic or manually created extraction rules. Traditional parsing methods often rely on fixed templates, whereas an LLM can accommodate a wider variety of invoice formats.

A limitation of this approach is that document quality directly affects extraction accuracy. If a PDF contains scanned images, handwritten text, or complex table layouts, additional OCR preprocessing may be required. Furthermore, generated JSON should always be validated before being used in production systems to ensure data integrity.
